### W8A1
[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao
(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện
tích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và
chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử
dụng lại những hàm vừa viết, π=3.14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 


```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
    def getArea(self):
        return 2*3.14*(self.r**2) + 2*(self.r*3.14)*self.h
    def getVolume(self):
        return 3.14*(self.r**2)*self.h
    def __str__(self):
        return f"{self.r}, {self.h}"
b, c= map(int,input().split())
ket_qua= Cylinder(b, c)
print(ket_qua)
print(f"({ket_qua.r},{ket_qua.h})")
print(f"{ket_qua.getArea():.2f}")
print(f"{ket_qua.getVolume():.2f}")
```

### W8A2
[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class MyDate:
    def __init__(self,dd,mm,yyyy):
        self.day = dd
        self.month = mm
        self.year = yyyy
    def checkValid(self):
        if self.year % 400 == 0 or (self.year % 4 == 0 and self.year % 100 != 0) :
            ngay_trong_thang = [31,29,31,30,31,30,31,31,30,31,30,31]
        else:
            ngay_trong_thang = [31,28,31,30,31,30,31,31,30,31,30,31]
        if not (1 <= self.month <= 12) or not (1 <= self.day <= ngay_trong_thang[self.month - 1]):
            return "INVALID"
        if self.month == 12 and self.day == 31:
            return f"1/1/{self.year + 1}"
        if self.day < ngay_trong_thang[self.month-1]:
            return f"{self.day + 1}/{self.month}/{self.year}"
        else:
            return f"1/{self.month + 1}/{self.year}"
d, m, y = map(int,input().split("/"))
ket_qua = MyDate(d,m,y)
print(ket_qua.checkValid())
```

### W8A3
[Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng các phương
thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục
tung không, có nằm trên trục hoành không, tính khoảng cách từ
điểm đó tới một điểm khác
Đầu vào:
Nhập vào tọa độ một điểm M bất kỳ
Đầu ra:
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ
là bao nhiêu (lấy 2 chữ số phần thập phân)

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def GocToaDo(self):
        return self.x == 0 and self.y == 0
    def ThuocTrucTung(self):
        return self.x == 0 and self.y != 0
    def ThuocTrucHoanh(self):
        return self.x != 0 and self.y == 0
    def tinhKhoangCach(self, other):
        khoangcach = ((self.x - other.x)**2 + (self.y - other.y)**2)**0.5
        return f"{khoangcach:.2f}"
x, y = map(int,input().split())
a, b = map(int,input().split())
P1 = Point(x,y)
P2 = Point(a,b)
print(P1.GocToaDo())
print(P1.ThuocTrucTung())
print(P1.ThuocTrucHoanh())
print(P2.tinhKhoangCach(P1))
```

### W8A4
[Calculator]
Xây dựng lớp Calculator (máy tính bỏ túi) với các phương thức
để thực hiện các phép toán 1. add (cộng), 2. subtract(trừ), 3.
product (nhân) và 4. divide(chia), 5. power(hàm mũ), 6. mod
(lấy phần dư) và phương thức 7. set_numbers(cập nhật 2 số
mới) nếu người dùng muốn tính toán trên 2 số khác.
Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương
thức thực hiện một phép tính và trả về kết quả tương ứng
Tạo menu và cho người dùng tính toán tới khi gõ "yes", "YES"
hoặc "Yes" cho câu hỏi "Bạn có muốn thoát?" mới dừng. 

```python
class Calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def subtract(self):
        return self.a - self.b
    def product(self):
        return self.a * self.b
    def divide(self):
        if self.b == 0:
            return "Loi chia cho 0"
        return self.a / self.b
    def power(self):
        return self.a ** self.b
    def mod(self):
        if self.b == 0:
            return "Loi chia cho 0"
        return self.a % self.b
    def set_number(self, new_a, new_b):
        self.a = new_a
        self.b = new_b
calc = Calculator(0, 0)
while True:
    print("\n--- MENU MÁY TÍNH ---")
    print("1. Nhập/Cập nhật 2 số")
    print("2. Tổng | 3. Hiệu | 4. Tích | 5. Thương | 6. Lũy thừa | 7. Chia dư")
    print("Nhập Yes để thoát.")
    choice = input("Lựa chọn của bạn: ").strip().lower()
    if choice == "yes":
        print("Tạm biệt!")
        break
    if choice == '1':
        a = float(input("Nhập số a: "))
        b = float(input("Nhập số b: "))
        calc.set_number(a, b)
        print(f"Đã cập nhật: a = {a}, b = {b}")
    elif choice == '2': print("Kết quả:", calc.add())
    elif choice == '3': print("Kết quả:", calc.subtract())
    elif choice == '4': print("Kết quả:", calc.product())
    elif choice == '5': print("Kết quả:", calc.divide())
    elif choice == '6': print("Kết quả:", calc.power())
    elif choice == '7': print("Kết quả:", calc.mod())
    else:
        print("Lựa chọn không hợp lệ, vui lòng thử lại.")
        
```

Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi
sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7.
Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có

```python
class ShoppingCart:
    def __init__(self):
        self.gio_hang = None
    def tao_dsach(self):
        self.gio_hang = []
        return "Đã tạo giỏ hàng"
    def them_san_pham(self, ten, gia):
        if self.gio_hang is None:
            print("Lỗi: Phải nhấn phím 1 để tạo giỏ hàng trước!")
            return
        san_pham = (ten,gia)
        self.gio_hang.append(san_pham)
        print(f"Đã thêm {ten} vào giỏ.")
    def xoa_san_pham(self, san_pham_can_xoa):
        if self.gio_hang is None:
            print("Lỗi: Phải nhấn phím 1 để tạo giỏ hàng trước!")
            return
        for i in range(len(self.gio_hang)):
            if self.gio_hang[i][0] == san_pham_can_xoa:
                ten_can_xoa = self.gio_hang.pop(i)
                print(f"Đã xoá sản phẩm")
                return
        print("Không tìm thấy sản phẩm")
    def kiem_tra_gio(self):
        if self.gio_hang is None:
            print("Lỗi: Phải nhấn phím 1 để tạo giỏ hàng trước!")
            return
        print(f"Giỏ có {len(self.gio_hang)} sản phẩm")
    def tong_tien(self):
        if self.gio_hang is None:
            print("Lỗi: Phải nhấn phím 1 để tạo giỏ hàng trước!")
            return
        tong = 0
        for san_pham in self.gio_hang:
            tong += san_pham[1]
        return tong
    def in_ten(self):
        if self.gio_hang is None:
            print("Lỗi: Phải nhấn phím 1 để tạo giỏ hàng trước!")
            return
        for san_pham in self.gio_hang:
            print(f"Tên :{san_pham[0]}. Giá: {san_pham[1]}")
    def xoa_gio_hang(self):
        gio_moi = []
        self.gio_hang = gio_moi
        print("Bạn đã xoá tất cả sản phẩm")
cart = ShoppingCart()
while True:
    print("\n--- MENU ---")
    choice = input("Lựa chọn của bạn: ").strip().lower()
    if choice == "yes":
        print("Tạm biệt!")
        break
    if choice == '1': print("Kết quả:", cart.tao_dsach())
    elif choice == '2': 
        ten = input("Nhập tên SP: ")
        try:
            gia = float(input("Nhập giá SP: "))
            cart.them_san_pham(ten, gia) 
        except ValueError:
            print("Lỗi: Giá phải là một con số!")
    elif choice == '3':
        ten_xoa = input("Nhập tên cần xóa: ") 
        cart.xoa_san_pham(ten_xoa)
    elif choice == '4': cart.kiem_tra_gio()
    elif choice == '5': print("Kết quả:", cart.tong_tien())
    elif choice == '6': cart.in_ten()
    elif choice == '7': cart.xoa_gio_hang()
    else:
        print("Lựa chọn không hợp lệ, vui lòng thử lại.")
```

### W8A6
Yêu cầu
Xây dựng lớp Rectangle(w, h) với các phương thức:
area() trả về diện tích.
perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h),
không trả về gì.
Input
w h k (số thực dương)
Output
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân)

```python
class Rectangle:
    def __init__(self,w,h):
        self.w = w
        self.h = h
    def area(self):
        return f"{self.w * self.h :.2f}"
    def perimeter(self):
        return f"{2 * (self.w + self.h):.2f}"
    def scale(self, k):
        self.w *= k
        self.h *= k
        return self
w, h, k = map(int,input().split())
hcn = Rectangle(w,h)
result = hcn.scale(k)
print(result.area())
print(result.perimeter())
```
