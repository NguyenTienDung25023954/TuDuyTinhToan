Viết hàm truyền vào hai số nguyên a, b. Trả về số lớn hơn trong hai số. def Max_Of_Two( a, b) -> 


```python
def max_of_two(a,b):
    if a > b:
        return a
    return b
a, b = map(int,input().split())
print(max_of_two(a,b))
```

### W5A3 Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 


```python
def so_nguyen_to(a):
    for i in range(2,int(a**0.5)+1):
        if a%i==0:
            return False
    else:
        return True
a=int(input())
print(so_nguyen_to(a))
```

### W5A4 VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên. 


```python
def so_hoan_hao(a):
    tong=0
    for i in range(1,int(a**0.5)+1):
        if a%i==0:
            tong+=i
            j=a//i
            if a%j==0 and j!=a:
                tong+=j
    if tong==a:
        return True
    else:
        return False
a=int(input())
print(so_hoan_hao(a))
```

### W5A6 Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!)

### W5A7
Viết hàm giả lập một máy tính bỏ túi.
Hàm thực hiện tính toán kết quả của phép toán giữa số \( num1 \) và \( num2 \) với toán tử \( operat \)
##### Biết các toán tử đơn giản thực hiện phép toán lần lượt là:
##### Phép trừ \( - \)
##### Phép cộng \( + \)
##### Phép chia \( / \)
##### Phép nhân \( * \)
##### Kết quả làm tròn đến chữ số thập phân thứ hai.
##### Đầu vào: Hai số thực và phép toán cần thực hiện (trong số các phép toán \(+, -, *, /\)).
##### Đầu ra: Kết quả phép toán

```python
def tinh_toan(num1,num2,operat):
    if operat == "+":
        return num1 + num2
    elif operat == "-":
        return num1 - num2
    elif operat == "*":
        return num1 * num2
    elif operat == "/":
        if num2==0:
            return False
        else:
            return num1 / num2
num1,num2=map(int,input().split())
operat=str(input())
print(tinh_toan(num1,num2,operat))
```

### W5A8 Cho hai số nguyên x và y, hãy viết hàm trả về khoảng cách Hamming giữa hai số này. 


```python
def nhi_phan(a):
    so_du=[]
    while a!=0:
        i=a%2
        so_du.append(i)
        a//=2
    so_du.reverse()
    nhiphan=int("".join(map(str, so_du)))
    return nhiphan
x, y= map(int,input().split())
m=str(nhi_phan(x))
n=str(nhi_phan(y))
if len(m) >= len(n):
    print(len(m)-len(n))
else :
    print(len(n)-len(m))
```

### W5A9 Viết hàm trả về tổng các chữ số của số nguyên dương n

```python
def tong_chu_so(n):
    tong=0
    while n%10!=0:
        i=n%10
        tong+=i
        n//=10
    return tong
n=int(input())
print(tong_chu_so(n))
```

### Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). 
Bạn có thể giả sử hai từ đầu vào có cùng độ dài. 

```python
def dang_cau(a,b):
    anh_xa_a_den_b={}
    anh_xa_b_den_a={}
    for i in range(len(a)):
        ki_tu_a=a[i]
        ki_tu_b=b[i]
    #Dictionary[Khóa] = Giá trị
        if ki_tu_a in anh_xa_a_den_b:
            if anh_xa_a_den_b[ki_tu_a] != ki_tu_b:
                return False
        else:
            anh_xa_a_den_b[ki_tu_a] = ki_tu_b
        if ki_tu_a in anh_xa_b_den_a:
            if anh_xa_a_den_b[ki_tu_b] != ki_tu_a:
                return False
        else:
            anh_xa_a_den_b[ki_tu_b] = ki_tu_a
    return True 
a=str(input())
b=str(input())
print(dang_cau(a,b))
```

### W5A11 Viết hàm is_product_divisible_by_10(a, b) trả về True nếu tích a*b chia hết cho 10, ngược lại False

### W5A12
Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor*q + r và 0 <= |r| < |divisor|.

```python
def divide_with_remainder(dividend, divisor):
    if divisor==0:
        return False
    q=dividend//divisor
    r=dividend-(divisor*q)
    return(q,r)
dividend, divisor=map(int,input().split())
print(divide_with_remainder(dividend, divisor))
```

### W5A15 Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông.

```python
def is_right_triangle(a, b, c):
    if a+b>c and a+c>b and b+c>a:
        return True
    else:
        return False
a,b,c=map(int,input().split())
print(is_right_triangle(a, b, c))
```
