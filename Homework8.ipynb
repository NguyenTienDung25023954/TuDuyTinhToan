### W8A16 
Yêu cầu
Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
###### Input
###### Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
###### Dòng 2: tham số tương ứng.
###### Output
###### In ra diện tích với 2 chữ số thập phân. 


```python
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * (self.r ** 2)
shape_type = str(input()).upper()
if shape_type == "RECTANGLE":
    w, h = map(float, input().split())
    shape = Rectangle(w, h)
elif shape_type == "CIRCLE":
    r = float(input())
    shape = Circle(r)
print(f"{shape.area():.2f}")
```

### W8A17 
Yêu cầu
Xây dựng lớp Vehicle(brand, speed).
Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.
Viết phương thức info() cho mỗi lớp:
- Vehicle: brand-speed
- Car: brand-speed-seats
###### Input
###### Dòng 1: loại đối tượng (Vehicle hoặc Car)
###### Dòng 2: dữ liệu tương ứng.
###### Output
###### In ra chuỗi info(). 

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand 
        self.speed = speed
    def info(self):
       return f"{self.brand}-{self.speed}"
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
    def info(self):
        return f"{super().info()}-{self.seats}"
type = str(input()).upper()
if type == "VEHICLE":
    brand, speed = map(str, input().split())
    result =  Vehicle(brand, speed)
elif type == "CAR":
    brand, speed, seats = map(str, input().split())
    result = Car(brand, speed, seats)
print(result.info())
```

### W8A18 
Yêu cầu
Xây dựng lớp Polygon với phương thức perimeter().
Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.
###### Input
###### Một dòng chứa 3 số nguyên a b c.
###### Output
###### In ra chu vi tam giác. 

```python
class Polygon:
    def perimeter(self):
        pass
class Triangle(Polygon):  # kế thừa
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):  # ghi đè
        return self.a + self.b + self.c
a, b, c = map(int, input().split())
result = Triangle(a,b,c)
print(result.perimeter())
```

### W8A19 
Yêu cầu
Xây dựng lớp Person(name, age) và lớp Student(name, age, school) kế thừa từ Person.
Phương thức describe():
- Person: name age
- Student: name age school
###### Input
###### Dòng 1: Person hoặc Student
###### Dòng 2: dữ liệu tương ứng.
###### Output
###### In ra kết quả describe().

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def describe(self):
        return f"{self.name}-{self.age}"
class Student(Person):
    def __init__(self,name,age,school):
        super().__init__(name,age)
        self.school = school
    def describe(self):
        return f"{super().describe()}-{self.school}"
type = str(input()).upper()
if type == "PERSON":
    name,age = map(str, input().split())
    result = Person(name,age)
elif type == "STUDENT":
    name,age,school = map(str, input().split())
    result = Student(name,age,school)
print(result.describe())
```
