### W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1


```python
def tim_vi_tri_gia_tri(dsach, k):
    m = len(dsach)
    for i in range(m):
        if dsach[i] == k:
            return i+1  # theo thứ tự 1,2,3
    return -1
dsach = list(map(int,input().split()))
k = int(input())
print(tim_vi_tri_gia_tri(dsach,k))
```

##### Vẫn yêu cầu đó nhưng tìm vị trí phần tử lớn hơn (bé hơn)

```python
def tim_vi_tri_gia_tri(dsach, k):
    m = len(dsach)
    vtri_so_thoa_man = []
    for i in range(m):
        if dsach[i] == k:
            vtri_so_thoa_man.append(i+1)  # theo thứ từ 1,2,3,...
    if len(vtri_so_thoa_man) == 0:
        return -1
    return max(vtri_so_thoa_man) # or min(vtri_so_thoa_man)
dsach = list(map(int,input().split()))
k = int(input())
print(tim_vi_tri_gia_tri(dsach,k))
```

```python
def tim_vi_tri_cuoi_cung(dsach, k):
    # Duyệt ngược từ len(dsach)-1 về 0 để tìm vị trí lớn nhất
    for i in range(len(dsach) - 1, -1, -1):
        if dsach[i] == k:
            return i + 1  # Trả về vị trí tự nhiên và thoát luôn
    return -1  # Duyệt hết mà không thấy thì mới trả về -1
```

### W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
dsach = list(map(str,input().split()))
ket_qua={}
for i,chuoi in enumerate(dsach): #lấy cả index và value
    if chuoi not in ket_qua:
        ket_qua[chuoi] = [i]
    else:
        ket_qua[chuoi].append(i)
for chuoi in ket_qua:
    ket_qua[chuoi] = tuple(ket_qua[chuoi]) #chuyển thành tuple
print(ket_qua)  #kết quả có (3,) để nhận biết dấu hiệu tuple
```

### W6A23 
Viết hàm xóa đi các số <x có trong mảng
###### Để thực hiện việc "xóa" trong lập trình, thực tế chúng ta thường chọn cách "giữ lại những gì thỏa mãn" rồi tạo ra mảng mới.

```python
def dsach_sau_khi_xoa(dsach,x):
    dsach_sau_khi_xoa=[]
    for so in dsach:
        if so < x :
            dsach_sau_khi_xoa.append(so)
    return dsach_sau_khi_xoa
dsach=list(map(int,input().split()))
x=int(input())
print(dsach_sau_khi_xoa(dsach,x))
```

### W6A24
Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def di_chuyen_phan_tu_0(dsach):
    dsach1=[]
    dsach2=[]
    for so in dsach:
        if so == 0:
            dsach1.append(so)
        else :
            dsach2.append(so)
    return dsach1+dsach2
dsach=list(map(int,input().split()))
print(di_chuyen_phan_tu_0(dsach))
```

### W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối
###### Một phần tử được gọi là cực đại địa phương nếu nó lớn hơn cả hai phần tử đứng sát cạnh nó (bên trái và bên phải)

```python
dsach = list(map(int,input().split()))
tong=0
n=len(dsach)
for i in range(1,n-1):
    if dsach[i] > dsach[i-1] and dsach[i] > dsach[i+1]:
        tong += dsach[i]
print(tong)
```

### W6A27
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1

```python
def tim_vtri(dsach,x):
    vi_tri=[]
    for i,so in enumerate(dsach):
        if so < x :
            vi_tri.append(i+1) #thứ tự 1,2,3,...
    return vi_tri
dsach=list(map(int,input().split()))
x=int(input())
print(tim_vtri(dsach,x))
```

### W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
def kiem_tra_day(dsach):
    n=len(dsach)
    if n < 2: return "Tăng thật sự"
    for i in range(1,n-1):
        if dsach[i-1]==dsach[i] or dsach[i]==dsach[i+1]:
            return "Vô danh"
        elif dsach[i] > dsach[i-1] and dsach[i] > dsach[i+1]:
            return "Vô danh"
        elif dsach[i] < dsach[i-1] and dsach[i] < dsach[i+1]:
            return "Vô danh"
    if dsach[0] < dsach[1]:
        return "Tăng thật sự"
    else:
        return "Giảm thật sự"
dsach = list(map(int,input().split()))
print(kiem_tra_day(dsach))
```
