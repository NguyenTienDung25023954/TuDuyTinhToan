### W7A31
Đổi chỗ đúng 2 chữ số để được số nhỏ
nhất có thể

```python
def doi_so_min(so):
    N = list(so.strip())
    n= len(N)
    for i in range(n-1):  
        vtri_min = i
        for j in range(n-1,i,-1):
            if vtri_min == i or N[j] < N[vtri_min]:
                vtri_min = j
        if vtri_min != i:
            N[i], N[vtri_min] = N[vtri_min], N[i]
            return "".join(N)
    N[-1], N[-2] = N[-2], N[-1]
    return "".join(N)
so = str(input())
print(doi_so_min(so))
```
