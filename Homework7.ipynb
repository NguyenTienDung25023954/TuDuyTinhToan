### W7A16
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp

```python
def thong_ke(arr):
    if not arr:
        return None
    ans = {}
    for i in arr:
        if i in ans:
            ans[i] += 1
        else:
            ans[i] = 1
    return ans
def mang_dep(arr):
    if not arr:
        return None
    ans = thong_ke(arr)
    y = 0
    for x, count in ans.items():
        if x == 0:
            y += count
        elif count >= x:
            y += (count - x)
        else:
            y += count
    ket_qua = []
    for x, count in ans.items():
        if x > 0 and count >= x:
            for _ in range(x):
                ket_qua.append(x)
    return y, ket_qua
arr = list(map(int,input().split()))
print(mang_dep(arr))
```

### W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
def tim_ptu_da_so(arr):
    n = len(arr)
    if n == 0:
        return None
    ung_vien = None
    count = 0
    for x in arr:
        if count == 0:
            ung_vien = x   #đổi ứng viên
            count = 1
        elif ung_vien == x:
            count += 1
        else :
            count -= 1
    actual_count = 0
    for x in arr:
        if x == ung_vien:
            actual_count += 1
    if actual_count > n / 2:
        return ung_vien
arr = list(map(int, input().split()))
print(tim_ptu_da_so(arr))
```

### W7A18
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần

```python
def sap_xep_co_dieu_kien(arr):
    n = len(arr)
    if n <= 1:
        return arr
    L = 0
    R = n - 1
    while L <= R:
        if arr[L] % 2 == 0:
            L += 1
        elif arr[R] % 2 != 0:
            R -= 1
        else:
            arr[L], arr[R] = arr[R], arr[L] #đổi chỗ
            L += 1
            R -= 1
     # Đoạn chẵn (từ đầu đến R) tăng dần
    arr[0:R+1] = sorted(arr[0:R+1])
    # Đoạn lẻ (từ L đến hết) giảm dần
    arr[L:n] = sorted(arr[L:n], reverse=True)
    return arr
arr = list(map(int,input().split()))
print(sap_xep_co_dieu_kien(arr))
```

### W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
def lay_cau_hinh_mang(arr): #vtri bắt đầu, bước nhảy, số phần tử còn lại
    n = len(arr)
    if n <= 1:
        return 0, 1, n
    if arr[0] <= arr[-1]:
        return 0, 1, n        # Mảng tăng 
    else:
        return n - 1, -1, n
def merge_hai_mang(A, B):
    pA, stepA, remA = lay_cau_hinh_mang(A)
    pB, stepB, remB = lay_cau_hinh_mang(B)
    ket_qua = []
    while remA > 0 and remB > 0:
        if A[pA] <= B[pB]:
            ket_qua.append(A[pA])
            pA += stepA
            remA -= 1
        else:
            ket_qua.append(B[pB])
            pB += stepB
            remB -= 1
    while remA > 0:
        ket_qua.append(A[pA])
        pA += stepA
        remA -= 1                   # Nếu mảng A/B còn phần tử, vét hết vào kết quả
    while remB > 0:
        ket_qua.append(B[pB])
        pB += stepB
        remB -= 1
    return ket_qua
A = list(map(int,input().split()))
B = list(map(int,input().split()))
print(merge_hai_mang(A,B))
```

### W7A20
Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi
N số nguyên dương, dòng sau ghi số nguyên dương k

```python
def tong_ptu_k_chu_so(arr,k):
    if not arr:
        return None
    tong = 0
    for N in arr:
        if 10**(k-1) <= N <= 10**k:  #tương đương N == len(str(N)) nhưng đỡ phức tạp hơn
            tong += N
    return tong
arr = list(map(int,input().split()))
k = int(input())
print(tong_ptu_k_chu_so(arr,k))
```

Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngày hợp lệ


```python
def tim_thu_tu_ngay(N,T): #năm 2020 là năm nhuận
    so_ngay_thang = [0, 31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if T < 1 or T > 12:
        return "invalid"
    if N < 1 or N > so_ngay_thang[T]:
        return "invalid"
    tong_ngay = 0
    for thang_truoc in range(1, T):
        tong_ngay += so_ngay_thang[thang_truoc]
    tong_ngay += N  #cộng số ngày tháng này
    return tong_ngay
N,T = map(int,input().split())
print(tim_thu_tu_ngay(N,T))
```

### W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất


```python
def tim_ten_goi_hay_gap_nhat(danh_sach_ho_ten):
    danh_sach_ten = {}
    for ho_ten in danh_sach_ho_ten:
        cac_tu = ho_ten.split()
        ten_goi = cac_tu[-1]
        if ten_goi in danh_sach_ten:
            danh_sach_ten[ten_goi] += 1
        else:
            danh_sach_ten[ten_goi] = 1
    ten_hay_gap = ""
    so_lan_max = 0
    for ten_goi, so_lan in danh_sach_ten.items():
        if so_lan > so_lan_max:
            so_lan_max = so_lan
            ten_hay_gap = ten_goi
        elif ten_hay_gap == "" or so_lan == so_lan_max:
            if ten_goi < ten_hay_gap:  #so trực tiếp chữ cái đầu
                ten_hay_gap = ten_goi
    return ten_hay_gap
N = int(input())
danh_sach = []
for i in range(N):
    danh_sach.append(input().strip())
print(tim_ten_goi_hay_gap_nhat(danh_sach))
```

### W7A24 
Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
def xoa_tu_lien_ke(xau):
    cac_tu = xau.split()
    if not cac_tu:
        return None
    ket_qua = [cac_tu[0]]
    for i in range(1, len(cac_tu)):# So sánh từ hiện tại với từ cuối cùng được giữ lại
        if cac_tu[i] != ket_qua[-1]:
            ket_qua.append(cac_tu[i])
    return " ".join(ket_qua)
xau = "An Bình Bình An đi đi học học"
print(xoa_tu_lien_ke(xau))
```

### W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
def tim_so_lon_nhat_co_k_chu_so(k,so):
    n = len(so)
    so_luot_xoa = n - k #dùng số lượt xoá để tiếp tục vòng lặp while cho chữ số đang xét
    ket_qua = []
    for chu_so in so:  #dùng len(ket_qua)>0 để tránh lỗi khi ktra ket_qua[-1] khi rỗng
        while len(ket_qua) > 0 and chu_so > ket_qua[-1] and so_luot_xoa > 0:
            ket_qua.pop() # Xóa chữ số nhỏ hơn ở phía trước
            so_luot_xoa -= 1 
        ket_qua.append(chu_so)
    so_lon_nhat = ket_qua[:k] #chỉ lấy k chứ số đầu
    return "".join(so_lon_nhat)
k = int(input())
so = str(input())
print(tim_so_lon_nhat_co_k_chu_so(k,so))
# dùng if thay while thì gặp dạng như 1119 sẽ sai
```

### W7A26
Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các
số đó. 

```python
def tach_va_tinh_so(van_ban):
    danh_sach_so = []
    so_tam = "" #gom các kí tự số vào số tạm
    n = len(van_ban)
    for i in range(n):
        ky_tu = van_ban[i]
        if ky_tu.isdigit():
            so_tam += ky_tu
        # Nếu là dấu chấm, phải kiểm tra xem có phải số thập phân không
        elif ky_tu == '.' and i > 0 and i < n - 1 and van_ban[i-1].isdigit() and van_ban[i+1].isdigit():
            so_tam += ky_tu   #nối ký tự dấu chấm . vào chuỗi chữ số đang gom dở
        else:
            if so_tam != "":
                danh_sach_so.append(float(so_tam))
                so_tam = "" # Xóa số tạm
    if so_tam != "":
        danh_sach_so.append(float(so_tam))
    if len(danh_sach_so) == 0:
        return 0, 0 
    tong = sum(danh_sach_so)
    trung_binh = tong / len(danh_sach_so)
    return tong, trung_binh
van_ban = input()
tong, trung_binh = tach_va_tinh_so(van_ban)
print(tong)
print(trung_binh)
#Học phí là 500k, điểm: 9.5
```

### W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
def kiem_tra_ngoai_hop_le(s):
    cap_ngoac = {')': '(', '}': '{', ']': '['}
    hang_cho = [] 
    for ky_tu in s:
        if ky_tu in cap_ngoac: # Nếu là ngoặc đóng trong dictionary
            if len(hang_cho) == 0 or hang_cho[-1] != cap_ngoac[ky_tu]:
                return False
            hang_cho.pop() # Khớp cặp thành công, xóa ngoặc mở đó khỏi hàng chờ
        else:
            hang_cho.append(ky_tu)
    # duyệt hết mà hàng chờ vẫn còn ngoặc mở dư thừa 
    return len(hang_cho) == 0
chuoi = input()
print(kiem_tra_ngoai_hop_le(chuoi)) 
```

### W7A28
Bài toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
def tim_so_nguyen_to(x):
    for i in range(2,int(x**0.5)+1):
        if x % i == 0:
            return False
    return True
def tim_day_hop_so(N):
    day_ky_luc = []
    day_hien_tai = []
    for i in range(1,N+2):
        la_hop_so = (i>1) and (not tim_so_nguyen_to(i))
        if i <= N and la_hop_so:
            day_hien_tai.append(i)
        else:
            if len(day_hien_tai) > len(day_ky_luc):
                day_ky_luc = day_hien_tai
            day_hien_tai = [] # Reset để tìm dãy tiếp theo
    return day_ky_luc
N = int(input())
print(tim_day_hop_so(N))
```

### W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau


```python
def tim_day_con_tang_lien_tuc(arr):
    if not arr:
        return []
    day_ky_luc = [arr] 
    day_hien_tai = [arr]
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            day_hien_tai.append(arr[i])
        else:
            if len(day_hien_tai) > len(day_ky_luc):
                day_ky_luc = day_hien_tai
            day_hien_tai = [arr[i]]
    if len(day_hien_tai) > len(day_ky_luc):
        day_ky_luc = day_hien_tai
    return day_ky_luc
arr = list(map(int, input().split()))
print(tim_day_con_tang_lien_tuc(arr))
```

### W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
def tim_day_con_tang_khong_lien_tuc(nums):
    if not nums: 
        return []
    n = len(nums)
    do_dai = [1] * n   #lưu độ dài max của dãy tăng cho mỗi phần tử
    vi_tri = [-1] * n  #lưu vị trí của phần tử đứng trước
    for i in range(n): 
        for j in range(i):
            if nums[i] > nums[j]:
                do_dai_cua_day = do_dai[j] + 1 
                if do_dai_cua_day >= do_dai[i]: 
                    do_dai[i] = do_dai_cua_day
                    vi_tri[i] = j
    do_dai_max = 0
    vi_tri_ket_thuc = 0
    for i in range(n):
        if do_dai[i] > do_dai_max:
            do_dai_max = do_dai[i]
            vi_tri_ket_thuc = i
    ket_qua = []
    vi_tri_dang_xet = vi_tri_ket_thuc  
    while vi_tri_dang_xet != -1:
        ket_qua.append(nums[vi_tri_dang_xet])
        vi_tri_dang_xet = vi_tri[vi_tri_dang_xet]
    return " ".join(map(str,ket_qua[::-1]))
nums = list(map(int,input().split()))
print(tim_day_con_tang_khong_lien_tuc(nums))
    
```
