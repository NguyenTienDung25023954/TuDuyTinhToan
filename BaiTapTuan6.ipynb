### W6A9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một
dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ
được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới
theo thứ tự từ điển tăng dần của key. 


```python
def nhap_dict():
    chuoi = input().split()
    dct = {}
    for item in chuoi:
        k, v = item.split(':')
        dct[k] = int(v)
    return dct
dict1 = nhap_dict()
dict2 = nhap_dict()    #sử dùng hàm để nhập dict
ket_qua = dict1.copy() #sử dụng bản sao của dict 1 (QUAN TRỌNG)
for k,v in dict2.items(): #trả về 1 tuple chứ cặp key:value
    if k in ket_qua:
        ket_qua[k] += v #cộng dồn nếu trùng key
    else :
        ket_qua[k] = v
dict_da_sap_xep={}
for k in sorted(ket_qua):
    dict_da_sap_xep[k] = ket_qua[k]
print(dict_da_sap_xep)
```

### W6A10
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên
dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là
một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh
sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được
sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple

```python
def tong_bang_k(dsach,k):
    n=len(dsach)
    ket_qua=[]
    for i in range(n):
        for j in range(i+1,n):
            if dsach[i]+dsach[j]==k:
                ket_qua.append((dsach[i],dsach[j]))
    ket_qua=list(set(ket_qua))
    ket_qua.sort(key=lambda x: x[0]) #sắp xếp tăng dần dựa trên phần tử đầu tiên
    return ket_qua
dsach= list(map(int,input().split()))
k=int(input())
print(tong_bang_k(dsach,k))
```

### W6A11
[Tách số chẵn và lẻ từ tuple]
Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa
tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên
thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.


```python
def tach_chan_le(tup):
    tup_le=[]
    tup_chan=[]  #dùng list để append
    for so in tup:
        if so %2 ==0:
            tup_chan.append(so)
        else:
            tup_le.append(so)
    return tuple(tup_le), tuple(tup_chan)  #ép tuple thành list
tup=tuple(map(int,input().split()))
result=tach_chan_le(tup)
print(result[0])
print(result[1])
```

### W6A12
[Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong
list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ
nhất trong số đó.


```python
def xuat_hien_nhieu_nhat(dsach):
    dem={}
    for so in dsach:
        if so in dem:
            dem[so]+=1
        else:
            dem[so]=1
    xuat_hien_nhieu_nhat=0
    so_xuat_hien_nhieu_nhat=[]
    for so in dem:
        if dem[so] > xuat_hien_nhieu_nhat:
            xuat_hien_nhieu_nhat = dem[so]
    for so in dem:
        if dem[so] == xuat_hien_nhieu_nhat:
            so_xuat_hien_nhieu_nhat.append(so)
    return min(so_xuat_hien_nhieu_nhat)
dsach=list(map(int,input().split()))
print(xuat_hien_nhieu_nhat(dsach))
```

### W6A13
[Đảo ngược key và value của dictionary]
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo
một dictionary mới bằng cách đảo ngược key và value của dictionary ban
đầu. In ra dictionary mới.


```python
def nhap_dict():
    chuoi = input().split()
    d={}
    for item in chuoi:
        k, v =item.split(":")
        d[k] = v
    return d
def dao_nguoc_key_value(dict1):
    dict2 = dict1.copy()
    dict_dao_ngc = {}
    for k, v in dict2.items():
        dict_dao_ngc[v] = k
    return dict_dao_ngc
dict1 = nhap_dict()
print(dao_nguoc_key_value(dict1))
```

### W6A14 [Giao của hai danh sách]

phần tử chung đó, không trùng lặp

### W6A15
[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp
key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary
mới.

```python
def nhap_dict():
    chuoi = input().split()
    d={}
    for item in chuoi:
        k, v = item.split(":")
        d[k] = int(v)  #thêm int để so sánh
    return d
def dict_co_value_lon_hon_k(dict1,k):
    ket_qua={}
    for key, value in dict1.items():
        if value > k:
            ket_qua[key] = value
    return ket_qua
dict1 = nhap_dict()
k = int(input())
print(dict_co_value_lon_hon_k(dict1,k))
```
