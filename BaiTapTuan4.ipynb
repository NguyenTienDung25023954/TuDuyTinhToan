### W4A7
"[LargestPrimeFactor]
Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n."

```python
def kiem_tra_so_nguyen_to(m):
    if m<2:
        return False
    for i in range(2,int(m**0.5)+1):
        if m%i==0:
            return False
    return True
n= int(input())
max_snt=0
for m in range(2,n+1):
    if n%m==0 and kiem_tra_so_nguyen_to(m):  #tìm ước rồi kiểm tra SNT
        if m > max_snt:
            max_snt=m
print(max_snt)
```

### W4A8
[ReverseAddPalindrome]
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 

```python
def palindrome(n):
    m = str(n)[::-1]
    buoc=0
    while m != m[::-1]:
        n += int(m)
        m = str(n)[::-1]
        buoc += 1
        if buoc > 1000:  # Tránh vòng lặp vô hạn
            return "Không tìm được palindrome sau 1000 bước"
    return f"{buoc} {n}"
n = int(input())
print(palindrome(n))
```

### W4A9
[PerfectSquaresDistinctDigits]
Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu cách). Nếu không có, in ra (no number).

```python
def in_so_chinh_phuong(n):
    i=1
    results=[] 
    while i*i <= n:
        k = i*i
        chu = str(k)
        if len(chu) == len(set(chu)): #dùng set() để loại bổ chữ giống nhau
            results.append(chu)
        i+=1 
    return results 
n=int(input())
ket_qua=in_so_chinh_phuong(n)
if not ket_qua :
    print("No number")
else:
    print(" ".join(ket_qua)) #join chỉ hoạt động với kiểu str
```

### W4A10
"[LongestCollatzUpToN]
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất."

```python
#Dãy Collatz (hay bài toán là một chuỗi số được tạo ra từ một số nguyên dương bất kỳ: nếu chẵn, chia đôi (n/2); nếu 
#lẻ, nhân ba cộng một (3n+1). Giả thuyết cho rằng mọi số nguyên dương cuối cùng đều sẽ về 1, tạo thành vòng lặp 4_2_1
def collatz(n):
    Lmax=0
    xmax=0
    for i in range(1,n+1):
        do_dai=1
        x=i #x để tính, i giữ nguyên
        while x!=1:
            if x%2==1:
                x=3*x+1
            else:
                x=x//2
            do_dai+=1
    if do_dai > Lmax:
        Lmax=do_dai
        xmax=i
        return f"({xmax} {Lmax})"
n=int(input())
print(collatz(n))
```

### W4A11
[EvenDivisors] Viết chương trình đếm số lượng ước số chẵn của số nguyên dương n với n < 10^6.

```python
def uoc_chan(n):
    count=0
    if n%2!=0:
        return 0
    else :
        for i in range(1,int(n**0.5)+1):
            if n%i==0:
                m=n//i
                if i%2==0:
                    count+=1
                if m%2==0 and m!=i:
                    count+=1
    return count
n=int(input())
print(uoc_chan(n))
```

### W4A13
[ClosePairNumber]
Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật:
Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại.
Ví dụ 220 và 284 là một cặp số thân thiết
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55,
110. Tổng của 11 ước này cho chúng ta 284
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng
vừa đúng bằng 220
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không.
Nếu có, in ra màn hình "true", ngược lại, in ra "false"

```python
def tong_uoc(n):
    tong=0
    for i in range(1,int(n**0.5)+1):
        if n%i==0:
            m=n//i
            tong+=i
            if m!=i and m!=n: #không cộng chính nó
                tong+=m
    return tong
a, b=map(int,input().split())
if tong_uoc(a)==b and tong_uoc(b)==a:
    print("True")
else :
    print("False")
```

### W4A14
[GCD]
Cho hai số m và n là hai số nguyên dương nhập từ bàn phím. Viết chương trình
tìm ước chung lớn nhất của hai số m và n. 

```python
def uoc_chung_max(m,n):   #thuật toán euclid
    while n!=0:
        r=m%n
        m,n=n,r
    return m
m,n=map(int,input().split())
print(uoc_chung_max(m,n))
```

### W4A15
[ChickenAndDog]
"Vừa gà vừa chó,
Bó lại cho tròn,
Ba mươi sáu con,
Một trăm chân chẵn"
Từ bài toán dân gian trên, mở rộng thành chương trình nhận đầu vào là tổng số
con và tổng số chân của gà và chó. Nếu tìm được số phù hợp, in ra số lượng gà và
số lượng chó. Ngược lại, in ra "invalid".

```python
def giai_toan(tong_con,tong_chan):
    if tong_chan%2!=0 or tong_chan < 2*tong_con or tong_chan > 4*tong_con:
        return "invalid"
    else:
        so_cho=(tong_chan-2*tong_con)//2
        so_ga=tong_con-so_cho
        if so_cho < 0 or so_ga < 0 or (so_cho * 4 + so_ga * 2) != tong_chan:
            return "invalid"
    return f"{so_ga} con ga va {so_cho} con cho"
tong_con, tong_chan= map(int, input().split())
print(giai_toan(tong_con,tong_chan))
```
