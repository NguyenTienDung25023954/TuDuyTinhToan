Viết hàm kiểm tra một số có phải số Amstrong hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x

```python
def is_Amstrong(x):
#là số mà tổng của các chữ số, 
# mỗi chữ số được nâng lên lũy thừa bằng 
# với số chữ số của chính nó, bằng chính số đó.
    i=x
    n=len(str(x))
    tong=0
    # tổng chữ số
    while x != 0:
        m=x%10
        tong += m**n
        x//=10
    if tong == i:
        return True
    else:
        return False
x=int(input())
print(is_Amstrong(x))
```

### W5A18 Viết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên

```python
def so_doi_xung(x):
    so_nguoc = 0
    m=x
    while x != 0:
        i = x%10
        x //= 10
        so_nguoc = so_nguoc*10 + i
    if so_nguoc == m:
        return True
    else:
        return False
x=int(input())
print(so_doi_xung(x))
```

### W5A19 Viết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.

```python
def so_ky_dieu(x):
    tong = 0
    for i in range(1,int(x**0.5)+1):
        if x%i==0:
            tong += i
            j = x//i
            if x%j==0 and j!=x:
                tong += j
    if tong == x:
        return True
    else:
        return False
x=int(input())
print(so_ky_dieu(x))
    
```

### W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.


```python
def detect_number_type(s):
    try:
        int(s)
        return 'int'
    except ValueError:
        pass
    
    # Nếu không phải int, thử chuyển sang số thực (float)
    try:
        float(s)
        return 'float'
    except ValueError:
        pass
    
    # Nếu cả hai đều thất bại
    return 'neither'
s=input()
print(detect_number_type(s))
```

### W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    # Tính diện tích của tam giác tạo bởi ba điểm
    area = 0.5 * abs(x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2))
    if area == 0:
        return True
    else:        
        return False
x1,y1,x2,y2,x3,y3=map(int,input().split())  
print(are_collinear(x1, y1, x2, y2, x3, y3))
```

### W5A25
Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def dien_tich(p1,p2,p3):
    return abs(p1[0]*(p2[1] - p3[1]) + p2[0]*(p3[1] - p1[1]) + p3[0]*(p1[1] - p2[1])) / 2.0
def kiem_tra(p, a, b, c):
    s_abc = dien_tich(a, b, c)
    s_pab = dien_tich(p, a, b)
    s_pbc = dien_tich(p, b, c)
    s_pca = dien_tich(p, c, a)
    return abs(s_abc - (s_pab + s_pbc + s_pca)) < 1e-9
def point_in_quadrilateral(quad,p):
    a, b, c, d = quad
    return kiem_tra(p, a, b, c) or kiem_tra(p, a, c, d)
quad=[]
for i in range(4):
    point = tuple(map(float,input().split()))
    quad.append(point)
px, py = map(float,input().split())
p=(px, py)
if point_in_quadrilateral(quad, p):
    print("=> Điểm nằm TRONG hoặc TRÊN BIÊN tứ giác.")
else:
    print("=> Điểm nằm NGOÀI tứ giác.")
```

### W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3)

```python
import math
def tinh_goc(A,B,C):
    ba = (A[0]-B[0],A[1]-B[1])
    bc = (C[0]-B[0],C[1]-B[1])
    tich_vo_huong=ba[0]* bc[0] + ba[1] * bc[1]
    do_dai_ba=math.sqrt(ba[0]**2 + ba[1]**2)
    do_dai_bc=math.sqrt(bc[0]**2 + bc[1]**2)
    if do_dai_bc == 0 or do_dai_bc == 0:
        return False
    cosin_goc = tich_vo_huong / (do_dai_ba*do_dai_bc)
    cosin_goc = max(-1.0, min(1.0, cosin_goc))
    goc_rad = math.acos(cosin_goc)
    goc_do = math.degrees(goc_rad)   #tinh ra radian roi doi sang do
    return goc_do
A = (0, 3)
B = (0, 0)
C = (4, 0)
print(tinh_goc(A,B,C))
```

### W5A29 Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min].


```python
def find_max_min(a, b, c):
    # Tìm Max
    max_val = a
    if b > max_val:
        max_val = b
    if c > max_val:
        max_val = c
    
    # Tìm Min
    min_val = a
    if b < min_val:
        min_val = b
    if c < min_val:
        min_val = c
        
    return [max_val, min_val]
```

### W5A30 Viết hàm tính hệ số nhị thức C(n, k).

```python
# C(n,k) = (n!) / (k! . (n-k)!)
def tinh_giai_thua(m):
    ket_qua = 1
    for i in range(2, m + 1):
        ket_qua *= i
    return ket_qua
def he_so_nhi_thuc(n, k):
    # Kiểm tra điều kiện hợp lệ
    if k < 0 or k > n:
        return 0
    tu_so = tinh_giai_thua(n)
    mau_so = tinh_giai_thua(k) * tinh_giai_thua(n - k)
    
    return tu_so // mau_so
n,k=map(int,input().split())
print(he_so_nhi_thuc(n,k))
```

### W5A33. Yêu cầu: Cước phí Taxi Thông Minh.
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả,
kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 


```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type==4:
        if km<=0.8:
            gia_xe=11000
        elif 0.8<km<=20:
            gia_xe=11000+(km-0.8)*13000
        else:
            gia_xe=11000+(20-0.8)*13000+(km-20)*10500
    if car_type==7:
        if km<=0.8:
            gia_xe=12000
        elif 0.8<km<=20:
            gia_xe=12000+(km-0.8)*14000
        else:
            gia_xe=12000+(20-0.8)*14000+(km-20)*12000
    if wait_minutes<=5:
        phi_cho=0
    elif wait_minutes>5:
        phi_cho=((wait_minutes-5)//5+1)*2000
    total=gia_xe + phi_cho
    if start_hour in [22,23,0,1,2,3,4,5]:
        total=1.2*(gia_xe+phi_cho)
    else:
        if km > 30: 
            total*=0.9
    if total < 50000 and km > 5:
        total=50000  
    return total
car_type,start_hour,km,wait_minutes=map(float,input().split())
results=smart_taxi_fare(car_type, start_hour, km, wait_minutes)
print(int(results))
```

### W5A34
Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau. Nếu n ≤ 0 trả về 0.


```python
def chu_so_khac_nhau(x):
    s = str(abs(x))
    return len(s) == len(set(s))
def sum_1_to_n(n):
    if n <= 0:
        return 0
    tong = 0
    for x in range(1, n + 1):
        if chu_so_khac_nhau(x):
            tong += x  
    return tong
n=int(input())
print(sum_1_to_n(n))
```
