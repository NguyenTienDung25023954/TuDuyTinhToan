### W9A1
Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ
thuật toán sắp xếp nào 


```python
# vòng lặp while
def merge(lst1,lst2):
    n1 = len(lst1)
    n2 = len(lst2)
    i = 0
    j = 0 #nếu 2 dsach ngược chiều xếp thì tăng dần 0, giảm dần n-1
    dsach = []
    while i < n1 and j < n2:
        if lst1[i] <= lst2[j]:    
            dsach.append(lst1[i])
            i += 1
        else:
            dsach.append(lst2[j])
            j += 1
    if i < n1 :
        dsach.extend(lst1[i:])
    elif j < n2 :
        dsach.extend(lst2[j:])
    return dsach
lst1 = list(map(int,input().split()))
lst2 = list(map(int,input().split()))
print(" ".join(map(str,merge(lst1,lst2))))
```

```python
# đệ quy
def merge_2_dsach(lst1,lst2):
    if not lst1:
        return lst2
    if not lst2:
        return lst1
    # bước đệ quy
    if lst1[0] <= lst2[0]:
        return [lst1[0] + merge_2_dsach(lst1[1:],lst2)]
    else:
        return [lst2[0] + merge_2_dsach(lst1,lst2[1:])]
lst1 = list(map(int,input().split()))
lst2 = list(map(int,input().split()))
result=merge(lst1,lst2)
print(*result)
```

### W9A2
Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên
dòng tăng dần và trên cột tăng dần, hãy tìm số nhỏ thứ k (k <
nxm) với thuật toán tốt nhất có thể 


```python
def dem_phan_tu(matrix,mid,n,m):
    count = 0
    row = n-1
    col = 0
    while row >= 0 and col < m:
        if matrix[row][col] <= mid:
            count += (row + 1) #cộng tất cả bên trên
            col += 1 # tiến sang phải
        else:
            row -= 1  #lùi lên
    return count
def tim_so_nho_thu_k(matrix,k,n,m):
    low = matrix[0][0]
    high = matrix[n-1][m-1]
    ans = low
    while low <= high:
        mid = (low + high) // 2
        if dem_phan_tu(matrix,mid,n,m) >= k:
            ans = mid
            high = mid - 1
        else:
            low = mid + 1
    return ans
n, m, k = map(int, input().split())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(tim_so_nho_thu_k(matrix,k,n,m))
```

### W9A3
Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ
số để được một số nguyên nhỏ nhất có thể 


```python
def doi_2_chu_so(so):
    N = list(so.strip())
    n = len(N)
    for i in range(n-1):
        vtri_min = i
        for j in range(n-1,i,-1):
            if N[j] < N[vtri_min] or vtri_min == i:
                vtri_min = j
        if vtri_min != i:
            N[i], N[vtri_min] = N[vtri_min], N[i]
            return "".join(N)
    N[-1], N[-2] = N[-2], N[-1]
    return "".join(N)
so = str(input())
print(doi_2_chu_so(so))
```

### W9A6
[Hoán vị tiếp theo]
Cho một dãy số gồm n chữ số là một hoán vị của dãy 1,2,...,n.
Viết chương trình in ra hoán vị ở vị trí tiếp theo (danh sách các
hoán vị được sắp xếp theo thứ tự từ điển) của hoán vị đã cho.
Đầu vào từ bàn phím gồm 2 dòng:
- Dòng đầu tiên chứa số nguyên n
- Dòng tiếp theo chứa n số nguyên là một hoán vị của dãy 1,2,...,n
(n≤105)
##### Đầu ra: In ra màn hình hoán vị tiếp theo của hoán vị đã cho. 


```python
def hoan_vi_tiep_theo(hoan_vi):
    n = len(hoan_vi)
    i = n - 2
    while i >= 0 and hoan_vi[i] >= hoan_vi[i+1]:
        i -= 1
    if i == -1:
        hoan_vi.reverse()
        return hoan_vi
    j = n - 1
    while j >= 0 and hoan_vi[j] <= hoan_vi[i]:
        j -= 1
    hoan_vi[j], hoan_vi[i] = hoan_vi[i], hoan_vi[j]
    hoan_vi[(i+1) :] = reversed(hoan_vi[(i+1) :])
    return hoan_vi
n = int(input())
hoan_vi = list(map(int,input().split()))
ket_qua=hoan_vi_tiep_theo(hoan_vi)
print(*ket_qua)
```

### W9A7
[CanPlaceFlowers]
Viết hàm canPlaceFlowers(flowerbed, k) kiểm tra xem có thể
trồng hết số hoa mới được cấp vào trong luống đất của bạn hay
không.
Hàm nhận đầu vào là một danh sách (list) flowerbed, đánh dấu
những vị trí trong luống đất đã được trồng cây xanh trước đó (giá
trị 1 ứng với vị trí cây đã được trồng và 0 ứng với vị trí đất trống)
và số lượng hoa được phát thêm k (int). Hàm trả về True nếu có
thể trồng hết toàn bộ k cây hoa vào luống đất và False trong
trường hợp ngược lại

```python
def canPlaceFlowers(flowerbed, k):
    if k == 0:
        return True
    n = len(flowerbed)
    for o in range(n):
        if flowerbed[o] == 0:
            trai_trong = (o == 0) or (flowerbed[o-1] == 0) 
            phai_trong = (o == n-1) or (flowerbed[o+1] == 0) 
            if trai_trong and phai_trong:
                k -= 1
                flowerbed[o] = 1 #quan trọng
            if k <= 0:
                return True
    return False
flowerbed = list(map(int,input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))
```

### W9A8
[Phát hiện đường đi lặp lại của Rô-bốt]
Ban đầu, một con rô-bốt ở vị trí (0, 0) trên một lưới 2D.
Cho một chuỗi `moves` chứa các thao tác di chuyển của rô-bốt.
Các bước di chuyển được biểu diễn bởi các ký tự:
- 'R' (Phải)
- 'L' (Trái)
- 'U' (Lên)
- 'D' (Xuống)
+ Viết hàm `def is_path_crossing(moves)`:
+ Hàm nhận đầu vào là một chuỗi `moves`.
+ Hàm trả về `True` nếu trong quá trình di chuyển, rô-bốt đi vào một ô mà nó đã từng đi qua trước đó.
+ Hàm trả về `False` nếu rô-bốt không bao giờ đi lặp lại (luôn đi vào ô mới).
+ Lưu ý: Vị trí ban đầu (0, 0) được tính là vị trí đầu tiên đã đi qua.

```python
def is_path_crossing(moves):
    da_di_qua = {(0,0)}
    x = 0
    y = 0
    for buoc in moves:
        if buoc == "R":
            x += 1
        elif buoc == "L":
            x -= 1
        elif buoc == "U":
            y += 1
        elif buoc == "D":
            y -= 1
        if (x,y) in da_di_qua:
            return True
        da_di_qua.add((x,y))
    return False
moves = input().strip()
print(is_path_crossing(moves))
```

### W9A9
[HARD - Tìm trung vị của 5 số]
Viết hàm `def Median(nums):`
Hàm nhận đầu vào là một danh sách (list) `nums` được đảm bảo
có đúng 5 số nguyên.
Yêu cầu:
Tìm và trả về số trung vị (số lớn thứ ba) của 5 số này.
Ràng buộc:
1. Chỉ được sử dụng tối đa 6 phép so sánh (ví dụ: <, >, <=, >=).
2. KHÔNG ĐƯỢC DÙNG bất kỳ hàm sắp xếp nào (như
`sorted()`, `list.sort()`).
Chú ý: Bạn chỉ cần viết hàm `Median` theo yêu cầu.
Phần đọc input và gọi hàm (hàm main) đã có sẵn. 

```python
def Median(nums):
    a,b,c,d,e=nums
    def doi_cho(x,y):
        if x > y:
            return (y,x)
        else:
            return (x,y)
    a,b = doi_cho(a,b)  #đảm bảo a < b
    c,d = doi_cho(c,d)  #đảm bảo c < d
    if b > d:
        a,b,c,d = c,d,a,b  #để b < d
    #'d' lớn hơn ít nhất 3 số (a, b, c) -> d không phải trung vị.
    #'a' nhỏ hơn ít nhất 2 số (b, d) -> a không phải trung vị.
    c,e = doi_cho(c,e)  #đảm bảo c < e
    if b < c:  #số nhỏ nhất là b -> bị loại
        return c #c < e -> b<c<e -> c là trung vị
    else:   #số nhỏ nhất là c -> bị loại
        if b < e:
            return b
        else:
            return e
nums = list(map(int,input().split()))
print(Median(nums))
```

### W9A10
[Tìm bán kính lò sưởi nhỏ nhất]
Mùa đông đang đến! Bạn được cung cấp vị trí của các ngôi nhà
và các lò sưởi trên một đường thẳng.
Nhiệm vụ của bạn là tìm bán kính r nhỏ nhất cho các lò sưởi sao
cho mọi ngôi nhà đều nằm trong phạm vi sưởi ấm của ít nhất một
lò sưởi.
Tất cả các lò sưởi đều có cùng một bán kính r.
+ Một ngôi nhà tại vị trí h được sưởi ấm bởi một lò sưởi tại vị trí t nếu |h - t| <= r.
+ Viết hàm `def findRadius(houses, heaters):`
+ Hàm nhận đầu vào là hai danh sách (list) các số nguyên:
- `houses`: chứa vị trí của các ngôi nhà.
- `heaters`: chứa vị trí của các lò sưởi.
+ Hàm trả về một số nguyên là bán kính r nhỏ nhất cần thiết.
+ Lưu ý:
- Số lượng nhà và lò sưởi là một số không âm và không vượt quá
25000.
- Vị trí của các ngôi nhà và lò sưởi là một số không âm và không
vượt quá 10^9.

```python
import bisect
def findRadius(nha, lo_suoi):
    lo_suoi.sort()
    nha.sort()
    ban_kinh_max = 0
    for house in nha:
        idx = bisect.bisect_left(lo_suoi, house)
        khoang_cach_phai = lo_suoi[idx] - house if idx < len(lo_suoi) else float('inf')
        khoang_cach_trai = house - lo_suoi[idx - 1] if idx > 0 else float('inf')
        khoang_cach_min = min(khoang_cach_phai, khoang_cach_trai)
        ban_kinh_max = max(ban_kinh_max, khoang_cach_min)
    return ban_kinh_max
nha = list(map(int,input().split()))
lo_suoi = list(map(int,input().split()))
print(findRadius(nha, lo_suoi))
```
