### W4A16 In các số chẵn chia hết cho 3 bé hơn 100

```python
for i in range(0,100,6):
    print(i, end=" ")
```

```python
for i in range(0,100):
    if i%2==0 and i%3==0:
        print(i, end=" ")
```

### W4A17 Nhập vào số nguyên dương a, in ra bảng cửu chương của a

```python
a=int(input())
for i in range(1,11):
    m=a*i
    print(f"{a} x {i} = {m}")
```

### W4A18 Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b 

```python
a, b=map(int, input().split())
for i in range(1,int(a**0.5)+1):
    if a%i==0 and b%i==0:
        print(i, end=" ")
        j=a//i
        if a%j==0 and b%j==0 and j!=i:
            print(j, end=" ")
```
